### Exercise 2: Count Total Nodes in a Binary Tree
* **Task:** Write a recursive function `count_nodes(root)` that counts and returns the total number of nodes in any binary tree.

```python
# Function that counts all nodes, starting from the root
def count_nodes(root): 
    # Base case: if there is no node here, stop the recursion 
    if root is None:  
        return 0 
   # Recursive case: count this node (1) and all nodes in the left and right subtrees
    return 1 + count_nodes(root.left) + count_nodes(root.right)  

# Test
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
root.right.left = Node(6)
root.right.right = Node(7)

# Call the function on the tree and print the total
print(f"Total number of nodes in the BST: {count_nodes(root)}")
# Expected output: 7
```

##  Exercise 4: Challenge — Build and Traverse a BST

Insert the following numbers into a Binary Search Tree:

**55, 35, 75, 25, 45, 65, 85, 15, 30, 40, 50, 60, 70, 80, 90**

### Tasks

1. Construct the BST by inserting the numbers **in the exact order given**.
2. Draw the completed tree.
3. Perform all four traversals:

   - Inorder
   - Preorder
   - Postorder
   - Level-Order

4. Write the result of each traversal.

### Reflection Questions

1. Which traversal produces the numbers in ascending order?
2. Which traversal visits the root first?
3. Which traversal visits the root last?
4. Which traversal uses a queue rather than following the depth-first pattern?
5. Which of the four traversals are considered **DFS traversals**?

# COMPLETED TREE
                                                                  55
                                                              /        \
                                                           35            75
                                                         /    \         /    \
                                                      25      45       65     85
                                                    /   \    /  \      /  \    / \
                                                  15    30   40  50   60  70  80  90                                       
                                                    
                                                

```python
# Build the BST
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None

def insert(root, val):
    if root is None:
        return Node(val)
    if val < root.val:
        root.left = insert(root.left, val)
    else:
        root.right = insert(root.right, val)
    return root

root = None
for n in [55, 35, 75, 25, 45, 65, 85, 15, 30, 40, 50, 60, 70, 80, 90]:
    root = insert(root, n)

# Implementing Tree Traversals
def inorder(root):
    if root:
        inorder(root.left)
        print(root.val, end=" ")
        inorder(root.right)

def preorder(root):
    if root:
        print(root.val, end=" ")
        preorder(root.left)
        preorder(root.right)

def postorder(root):
    if root:
        postorder(root.left)
        postorder(root.right)
        print(root.val, end=" ")

def levelorder(root):
    queue = [root]
    while queue:
        node = queue.pop(0)
        print(node.val, end=" ")
        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)

# --- Testing Traversals on our BST ---
print("Inorder Traversal (Sorted): ", end="")
inorder(root)
print()

print("Preorder Traversal: ", end="")
preorder(root)
print()

print("Postorder Traversal: ", end="")
postorder(root)
print()

print("Level-order Traversal: ", end="")
levelorder(root)
print()
```

# **Reflection answers*
* 1.Inorder gives ascending order. 
* 2.Preorder 
* 3.Postorder 
* 4.Level-order (BFS) uses a queue instead of following the depth-first pattern.
* 5.Inorder, preorder, and postorder are the DFS traversals. Level-order is BFS.
